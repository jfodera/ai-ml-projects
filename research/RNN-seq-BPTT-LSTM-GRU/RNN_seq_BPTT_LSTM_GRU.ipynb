## Intro

- NN's are typically designed in such a way that input features are largely independent of one another.
- But, some datatypes have sequential dependencies:
- *Time Series Data*
  - Values on successive time stamps are closely related to one another
  - This prevents us from deeming every successive value independent from one another
- *Textual Data*
  - Recently, there is a greater emphasis on the ordering of words being used for better semantic insights.
  - So, It is important to construct a model that takes sequential information into account
- *Biological Data*
  - Sequences corresponding to amino acids sequential ordering matters.

## Processing Sequences


- Recall: all deep learning models identify numbers not text or symbols. So, we need to represent these non-numeric values
- Consider the 'bag of words' model and the sentance "*Analystics is hardly any fun*"
- Assume we want to do sentiment analysis. A possible solution is to create one-hot encoded vectors for each of these words.
  - Each vector having a size of 20000 as that is the typical number of words used in English vocabulary  
- These vectors can be used as inputs to our feed forward network to perform sentiment analysis.
  - In input layer we can have one input for each word i nthe sentence
- Note: the entire dataset can have different sized sentences and **using variable input sizes is impossible in a conventinal neural network**.
  - Also, small differences in word ordering can lead to semantically different connatations
- THerefore, it is important to encode word ordering more directly within the architecture of the network
  - Basically, work around these constraints to make word ordering a vital part of your model because it is important.
